```python
import tkinter as tk
from tkinter import filedialog, messagebox
from PIL import Image, ImageTk
import cv2
import numpy as np
```

```python
class ImageProcessorGUI:
    def __init__(self, root):
        self.root = root
        self.root.title("Computer Vision Project")

        button_frame = tk.Frame(root)
        button_frame.pack(pady=10)

        buttons = [
            ("1. Upload Image", self.upload_image),
            ("2. Apply Salt & Pepper Noise", self.apply_noise),
            ("3. Remove Salt & Pepper Noise", self.remove_noise),
            ("4. Apply Gaussian Blur", self.apply_gaussian),
            ("5. Apply Gaussian Noise", self.apply_gaussian_noise),
            ("6. Thresholding Segmentation", self.apply_thresholding),
            ("7. Edge-Based Segmentation", self.apply_edge_segmentation),
            ("8. Region-Based Segmentation (Watershed)", self.apply_watershed),
            ("9. Apply Dilation", self.apply_dilation),
            ("10. Apply Erosion", self.apply_erosion),
            ("11. Apply Opening", self.apply_opening),
            ("12. Apply Closing", self.apply_closing),
            ("13. Apply OTSU Thresholding", self.otsu_thresholding),
            ("14. Apply Adaptive Thresholding", self.apply_adaptive_threshold),
            ("Save Processed Image", self.save_image)
        ]

        for index, (text, command) in enumerate(buttons):
            row = index // 4
            col = index % 4
            tk.Button(button_frame, text=text, width=25, command=command).grid(row=row, column=col, padx=5, pady=5)


        self.image = None
        self.processed_image = None



        self.noise_btn = tk.Button(root, text="2. Apply Salt & Pepper Noise", command=self.apply_noise)

        self.remove_noise_btn = tk.Button(root, text="3. Remove Salt & Pepper Noise", command=self.remove_noise)

        self.gaussian_btn = tk.Button(root, text="4. Apply Gaussian Blur", command=self.apply_gaussian)

        self.gaussian_noise_btn = tk.Button(root, text="5. Apply Gaussian Noise", command=self.apply_gaussian_noise)

        self.threshold_btn = tk.Button(root, text="6.Global Thresholding Segmentation", command=self.apply_thresholding)

        self.edge_btn = tk.Button(root, text="7. Edge-Based Segmentation", command=self.apply_edge_segmentation)

        self.watershed_btn = tk.Button(root, text="8. Region-Based Segmentation (Watershed)", command=self.apply_watershed)

        self.dilation_btn = tk.Button(root, text="9. Apply Dilation", command=self.apply_dilation)

        self.erosion_btn = tk.Button(root, text="10. Apply Erosion", command=self.apply_erosion)

        self.opening_btn = tk.Button(root, text="11. Apply Opening", command=self.apply_opening)

        self.closing_btn = tk.Button(root, text="12. Apply Closing", command=self.apply_closing)

        self.otsu_thresholding_btn = tk.Button(root, text="13. Apply  OTSU Thersholding", command=self.otsu_thresholding)

        self.adaptive_thresholding_btn = tk.Button(root, text="14. Apply Adaptive Thersholding  ", command=self.apply_adaptive_threshold)

        self.save_btn = tk.Button(root, text="Save Processed Image", command=self.save_image)

        self.image_label = tk.Label(root)
        self.image_label.pack()

        self.team_label = tk.Label(root, text="Team Members: Sondos Abu Baker, Shahinaz Mahmmoud, Shahd Waleed, Shiamaa Ahmed, Lobna Adel", fg="gray")
        self.team_label.pack(side=tk.BOTTOM, pady=5)

    def upload_image(self):
        path = filedialog.askopenfilename()
        if path:
            self.image = cv2.imread(path)
            self.processed_image = self.image.copy()
            self.display_image(self.image)

    def display_image(self, img):
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img_pil = Image.fromarray(img_rgb)
        img_tk = ImageTk.PhotoImage(img_pil)
        self.image_label.config(image=img_tk)
        self.image_label.image = img_tk

    def apply_noise(self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return
        salt_prob = 0.01
        pepper_prob = 0.02
        noisy_img = self.salt_pepper_noise(self.image.copy(), salt_prob, pepper_prob)
        self.processed_image = noisy_img
        self.display_image(noisy_img)

    def salt_pepper_noise(self, image, salt_prob=0.01, pepper_prob=0.01):
        noisy_img = image.copy()
        total_pixels = image.shape[0] * image.shape[1]

        num_salt = int(salt_prob * total_pixels)
        for _ in range(num_salt):
            i = np.random.randint(0, image.shape[0])
            j = np.random.randint(0, image.shape[1])
            noisy_img[i, j] = [255, 255, 255]

        num_pepper = int(pepper_prob * total_pixels)
        for _ in range(num_pepper):
            i = np.random.randint(0, image.shape[0])
            j = np.random.randint(0, image.shape[1])
            noisy_img[i, j] = [0, 0, 0]

        return noisy_img

    def remove_noise(self):
        if self.processed_image is None:
            messagebox.showwarning("Warning", "Please apply noise first.")
            return
        denoised_img = cv2.medianBlur(self.processed_image, 3)
        self.processed_image = denoised_img
        self.display_image(denoised_img)

    def apply_gaussian(self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return
        gaussian_img = cv2.GaussianBlur(self.image, (5, 5), 0)
        self.processed_image = gaussian_img
        self.display_image(gaussian_img)

    def apply_gaussian_noise(self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return
        mean = 0
        sigma = 25
        noise = np.random.normal(mean, sigma, self.image.shape)
        noisy_img = self.image.astype(np.float32) + noise
        noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)
        self.processed_image = noisy_img
        self.display_image(noisy_img)

    def apply_thresholding(self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return
        gray = cv2.cvtColor(self.image, cv2.COLOR_BGR2GRAY)
        _, thresh_img = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
        thresh_img = cv2.cvtColor(thresh_img, cv2.COLOR_GRAY2BGR)
        self.processed_image = thresh_img
        self.display_image(thresh_img)



    def apply_edge_segmentation(self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return


        gray = cv2.cvtColor(self.image, cv2.COLOR_BGR2GRAY)
        _, mask = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

        kernel = np.ones((3, 3), np.uint8)
        eroded = cv2.erode(mask, kernel, iterations=1)
        refined_edges = cv2.subtract(mask, eroded)

        refined_edges_bgr = cv2.cvtColor(refined_edges, cv2.COLOR_GRAY2BGR)
        self.processed_image = refined_edges_bgr
        self.display_image(refined_edges_bgr)

    def apply_watershed(self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return

        img = self.image.copy()
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        ret, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

        kernel = np.ones((3, 3), np.uint8)
        sure_bg = cv2.dilate(thresh, kernel, iterations=3)

        dist_transform = cv2.distanceTransform(thresh, cv2.DIST_L2, 5)
        ret, sure_fg = cv2.threshold(dist_transform, 0.7 * dist_transform.max(), 255, 0)

        sure_fg = np.uint8(sure_fg)
        unknown = cv2.subtract(sure_bg, sure_fg)

        ret, markers = cv2.connectedComponents(sure_fg)
        markers = markers + 1
        markers[unknown == 255] = 0

        markers = cv2.watershed(img, markers)
        img[markers == -1] = [255, 0, 0]

        self.processed_image = img
        self.display_image(img)

    def apply_dilation(self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return
        kernel = np.ones((5, 5), np.uint8)
        dilated = cv2.dilate(self.image, kernel, iterations=1)
        self.processed_image = dilated
        self.display_image(dilated)

    def apply_erosion(self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return
        kernel = np.ones((5, 5), np.uint8)
        eroded = cv2.erode(self.image, kernel, iterations=1)
        self.processed_image = eroded
        self.display_image(eroded)

    def apply_opening(self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return
        gray_img = cv2.cvtColor(self.image, cv2.COLOR_BGR2GRAY)
        resized_img = cv2.resize(gray_img, (256, 256))
        kernel = cv2.getStructuringElement(cv2.MORPH_CROSS, (5, 5))
        opening = cv2.morphologyEx(resized_img, cv2.MORPH_OPEN, kernel)
        cv2.putText(opening, "Opening", (10, 25), cv2.FONT_HERSHEY_SIMPLEX, 0.8, 255, 2)
        opening_colored = cv2.cvtColor(opening, cv2.COLOR_GRAY2BGR)
        self.processed_image = opening_colored
        self.display_image(opening_colored)

    def apply_closing(self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return
        gray_img = cv2.cvtColor(self.image, cv2.COLOR_BGR2GRAY)
        resized_img = cv2.resize(gray_img, (256, 256))
        kernel = cv2.getStructuringElement(cv2.MORPH_CROSS, (5, 5))
        closing = cv2.morphologyEx(resized_img, cv2.MORPH_CLOSE, kernel)
        cv2.putText(closing, "Closing", (10, 25), cv2.FONT_HERSHEY_SIMPLEX, 0.8, 255, 2)
        closing_colored = cv2.cvtColor(closing, cv2.COLOR_GRAY2BGR)
        self.processed_image = closing_colored
        self.display_image(closing_colored)





    def otsu_thresholding (self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return


        gray = cv2.cvtColor(self.image, cv2.COLOR_BGR2GRAY)
        _, otsu_thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
        self.processed_image = otsu_thresh
        otsu_bgr = cv2.cvtColor(otsu_thresh, cv2.COLOR_GRAY2BGR)
        self.display_image(otsu_bgr)

    def apply_adaptive_threshold(self):
        if self.image is None:
            messagebox.showwarning("Warning", "Please upload an image first.")
            return

        gray = cv2.cvtColor(self.image, cv2.COLOR_BGR2GRAY)
        adaptive_ths = cv2.adaptiveThreshold(gray, 255,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,99,2)
        self.processed_image = adaptive_ths
        adaptive_bgr = cv2.cvtColor(adaptive_ths, cv2.COLOR_GRAY2BGR)
        self.display_image(adaptive_bgr)

    def save_image(self):
        if self.processed_image is not None:
            file_path = filedialog.asksaveasfilename(defaultextension=".png",
                                                     filetypes=[("PNG files", "*.png"), ("JPEG files", "*.jpg")])
            if file_path:
                cv2.imwrite(file_path, self.processed_image)
                messagebox.showinfo("Saved", "Image saved successfully!")
        else:
            messagebox.showwarning("Warning", "No processed image to save.")


```

```python

    root = tk.Tk()
    app = ImageProcessorGUI(root)
    root.mainloop()

```
