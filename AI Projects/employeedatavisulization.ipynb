```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(20, 50))
```

```python

numerical_cols = ['Age', 'Job_Level', 'Monthly_Income', 'Hourly_Rate', 'Years_at_Company',
                  'Years_in_Current_Role', 'Years_Since_Last_Promotion', 'Work_Life_Balance',
                  'Job_Satisfaction', 'Performance_Rating', 'Training_Hours_Last_Year',
                  'Project_Count', 'Average_Hours_Worked_Per_Week', 'Absenteeism',
                  'Work_Environment_Satisfaction', 'Relationship_with_Manager',
                  'Job_Involvement', 'Distance_From_Home', 'Number_of_Companies_Worked']


for col in numerical_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True, bins=20, color='skyblue')
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.tight_layout()
    plt.show()
```

```python

categorical_cols = ['Gender', 'Marital_Status', 'Department', 'Job_Role', 'Overtime', 'Attrition']

for col in categorical_cols:
    plt.figure(figsize=(6,4))
    sns.countplot(data=df, x=col, palette="Set2")
    plt.title(f'Countplot of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.tight_layout()
    plt.show()
```

```python

for col_cat in categorical_cols:
    for col_num in numerical_cols:
        plt.figure(figsize=(6,4))
        sns.histplot(data=df, x=col_num, hue=col_cat, bins=20, kde=True, palette='Set2')
        plt.title(f'Histogram of {col_num} by {col_cat}')
        plt.xlabel(col_num)
        plt.ylabel('Count')
        plt.tight_layout()
        plt.show()
```

```python

for col1 in categorical_cols:
    for col2 in categorical_cols:
        if col1 != col2:
            plt.figure(figsize=(6,4))
            sns.countplot(data=df, x=col1, hue=col2, palette="Set2")
            plt.title(f'Countplot of {col1} by {col2}')
            plt.xlabel(col1)
            plt.ylabel('Count')
            plt.legend(title=col2)
            plt.tight_layout()
            plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(data=df, x='Age', hue='Attrition', kde=True, bins=20, palette='Set2')
plt.title('Histogram of Age by Attrition')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Department', hue='Attrition', palette="Set2")
plt.title('Countplot of Department by Attrition')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
grouped = df.groupby('Gender')['Attrition'].value_counts(normalize=True).unstack()
for idx in grouped.index:
    plt.figure(figsize=(4,4))
    grouped.loc[idx].plot.pie(autopct='%1.1f%%', startangle=90, colors=sns.color_palette('Set2'))
    plt.title(f'Attrition Distribution for Gender = {idx}')
    plt.ylabel('')
    plt.tight_layout()
    plt.show()
```
