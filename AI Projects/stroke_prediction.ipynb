# **HANDLE MISSING VALUES**

```python
# Check how many missing values in each column
print(df.isnull().sum())
```

```python
# List columns with missing values
missing_cols = df.columns[df.isnull().any()]
print("Columns with missing values:", missing_cols)

# Handle numeric columns
for col in missing_cols:
    if df[col].dtype in ['float64', 'int64']:
        df[col].fillna(df[col].median(), inplace=True)
    else:
        # For categorical/text columns, fill with 'Unknown'
        df[col].fillna('Unknown', inplace=True)

# Verify missing values handled
print(df.isnull().sum())
```

```python
# List of categorical/text columns in your dataset
text_cols=['Patient Name', 'Gender', 'Marital Status', 'Work Type', 'Residence Type',
 'Smoking Status', 'Alcohol Intake', 'Physical Activity',
 'Family History of Stroke', 'Dietary Habits',
  'Symptoms', 'Diagnosis']


# Initialize LabelEncoder
le = LabelEncoder()

# Encode each categorical column
for col in text_cols:
    if col in df.columns:
        df[col] = le.fit_transform(df[col].astype(str))  # Convert to string before encoding

# Encode 'Diagnosis' column separately for correlation calculation
if 'Diagnosis' in df.columns:
    df['Diagnosis'] = le.fit_transform(df['Diagnosis'])

# Verify encoding
df.head()
```

# **processed data**

```python
# Copy original DataFrame
# -------------------------
df_processed = df.copy()
df_processed.head()
```

# **DEFINE FEATURES AND TARGET**

```python
# Drop identifier columns from Features
cols_to_drop = ['Patient ID', 'Patient Name']
X = df_processed.drop(columns=cols_to_drop + ['Diagnosis'])  # Features
y = df_processed['Diagnosis']  # Target
```

# **ENCODE CATEGORICAL VARIABLES**

```python
# List of categorical/text columns to encode
text_cols=['Patient Name', 'Gender', 'Marital Status', 'Work Type', 'Residence Type',
 'Smoking Status', 'Alcohol Intake', 'Physical Activity',
 'Family History of Stroke', 'Dietary Habits',
 'Symptoms', 'Diagnosis']


le = LabelEncoder()
for col in text_cols:
    if col in X.columns:
        X[col] = le.fit_transform(X[col].astype(str))
        y = le.fit_transform(y)
```

```python
# Parse Cholesterol Levels into numeric columns
if 'Cholesterol Levels' in X.columns:
    X['HDL_Cholesterol'] = X['Cholesterol Levels'].apply(lambda x: int(x.split(',')[0].split(':')[1].strip()))
    X['LDL_Cholesterol'] = X['Cholesterol Levels'].apply(lambda x: int(x.split(',')[1].split(':')[1].strip()))
    X = X.drop('Cholesterol Levels', axis=1)
    X.head()
```

```python
# Now X and y are ready for Train/Test Split
X.head(), y[:5]
```

# **SPLIT DATA INTO TRAIN AND TEST**

```python
# Split the dataset: 70% train, 30% test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

# **Models**

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LinearRegression
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, precision_score, recall_score, f1_score, roc_curve, auc
```

```python
# -------------------------
# Random Forest Classifier
# -------------------------
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)

# -------------------------
# Support Vector Machine
# -------------------------
svm_model = SVC(kernel='rbf', probability=True, random_state=42)

# -------------------------
# Linear Regression (for numeric prediction of stroke probability)
# -------------------------
lr_model = LinearRegression()
```

```python
from sklearn.preprocessing import StandardScaler

# Ensure 'Blood Pressure Levels' is processed if it still exists in X_train/X_test
# Process X_train
if 'Blood Pressure Levels' in X_train.columns:
    X_train['Systolic_BP'] = X_train['Blood Pressure Levels'].apply(lambda x: int(x.split('/')[0].strip()))
    X_train['Diastolic_BP'] = X_train['Blood Pressure Levels'].apply(lambda x: int(x.split('/')[1].strip()))
    X_train = X_train.drop('Blood Pressure Levels', axis=1)

# Process X_test
if 'Blood Pressure Levels' in X_test.columns:
    X_test['Systolic_BP'] = X_test['Blood Pressure Levels'].apply(lambda x: int(x.split('/')[0].strip()))
    X_test['Diastolic_BP'] = X_test['Blood Pressure Levels'].apply(lambda x: int(x.split('/')[1].strip()))
    X_test = X_test.drop('Blood Pressure Levels', axis=1)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)  # Fit on train
X_test_scaled = scaler.transform(X_test)        # Apply same scaling to test
```

```python
# Train Random Forest
rf_model.fit(X_train_scaled, y_train)

```

```python
# Train SVM
svm_model.fit(X_train_scaled, y_train)

```

```python
# Train Linear Regression
lr_model.fit(X_train_scaled, y_train)
```

```python
# Random Forest
y_pred_rf = rf_model.predict(X_test_scaled)
```

```python
# SVM
y_pred_svm = svm_model.predict(X_test_scaled)
```

```python
# Linear Regression
y_pred_lr = lr_model.predict(X_test_scaled)
y_pred_lr_class = [1 if x>=0.5 else 0 for x in y_pred_lr]
```

```python
print("--- Random Forest ---")
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_rf))
print("Classification Report:\n", classification_report(y_test, y_pred_rf))
```

```python
print("--- Support Vector Machine (SVM) ---")
print("Accuracy:", accuracy_score(y_test, y_pred_svm))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_svm))
print("Classification Report:\n", classification_report(y_test, y_pred_svm))
```

```python
from sklearn.metrics import mean_squared_error

print("--- Linear Regression ---")
mse = mean_squared_error(y_test, y_pred_lr)
print("Mean Squared Error:", mse)
print("Rounded Classification Accuracy:", accuracy_score(y_test, y_pred_lr_class))
print("Classification Report:\n", classification_report(y_test, y_pred_lr_class))
```

# **Final Visualization**

```python
def plot_confusion_matrix(y_true, y_pred, model_name):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(5,4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title(f'Confusion Matrix - {model_name}')
    plt.ylabel('Actual')
    plt.xlabel('Predicted')
    plt.show()
```

```python
# Predictions
y_pred_rf = rf_model.predict(X_test_scaled)
y_pred_svm = svm_model.predict(X_test_scaled)
y_pred_lr = lr_model.predict(X_test_scaled)
y_pred_lr_class = [1 if x>=0.5 else 0 for x in y_pred_lr]
```

```python
def plot_confusion_matrix_interactive(y_true, y_pred, model_name):
    cm = confusion_matrix(y_true, y_pred)
    fig = px.imshow(
        cm,
        text_auto=True,
        labels=dict(x="Predicted", y="Actual", color="Count"),
        x=[0,1],
        y=[0,1],
        color_continuous_scale='Blues',
        title=f"{model_name} - Confusion Matrix (Interactive)"
    )
    fig.show()

# Interactive Confusion Matrices
plot_confusion_matrix_interactive(y_test, y_pred_rf, "Random Forest")
plot_confusion_matrix_interactive(y_test, y_pred_svm, "SVM")
plot_confusion_matrix_interactive(y_test, y_pred_lr_class, "Linear Regression")
```

```python
# Linear Regression coefficients
lr_coef = lr_model.coef_

# Random Forest feature importance
rf_importance = rf_model.feature_importances_

features = X_train.columns
df_feat = pd.DataFrame({
    'Feature': features,
    'LR_Coefficient': lr_coef,
    'RF_Importance': rf_importance
})

# Interactive heatmap
fig = px.imshow(
    df_feat.set_index('Feature'),
    text_auto=".2f",
    aspect="auto",
    color_continuous_scale='RdBu_r',
    title="Feature Importance - LR & RF (Interactive)"
)
fig.show()
```

```python
# Model Comparison
def model_metrics(y_true, y_pred, model_name):
    return {
        "Model": model_name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred),
        "Recall": recall_score(y_true, y_pred),
        "F1": f1_score(y_true, y_pred)
    }

metrics_list = [
    model_metrics(y_test, y_pred_rf, "Random Forest"),
    model_metrics(y_test, y_pred_svm, "SVM"),
    model_metrics(y_test, y_pred_lr_class, "Linear Regression")
]

df_metrics = pd.DataFrame(metrics_list)
print("Model Comparison:")
display(df_metrics)


best_model_name = df_metrics.sort_values(by='F1', ascending=False).iloc[0]['Model']
print(f"\nBest Model: {best_model_name}")
```

```python
# Model Comparison (re-included for robustness against NameError)
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def model_metrics(y_true, y_pred, model_name):
    return {
        "Model": model_name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, zero_division=1),
        "Recall": recall_score(y_true, y_pred, zero_division=1),
        "F1": f1_score(y_true, y_pred, zero_division=1)
    }

metrics_list = [
    model_metrics(y_test, y_pred_rf, "Random Forest"),
    model_metrics(y_test, y_pred_svm, "SVM"),
    model_metrics(y_test, y_pred_lr_class, "Linear Regression")
]

df_metrics = pd.DataFrame(metrics_list)
best_model_name = df_metrics.sort_values(by='F1', ascending=False).iloc[0]['Model']


# Example: New patient data (same columns and encoding as X_train)
new_data = pd.DataFrame({
    'Age':[65, 50, 72],
    'Gender':[1,0,1],
    'Hypertension':[1,0,1],
    'Heart Disease':[0,1,0],
    'Marital Status':[1,0,1],
    'Work Type':[2,1,0],
    'Residence Type':[1,0,1],
    'Average Glucose Level':[140, 110, 180],
    'Body Mass Index (BMI)':[28.5, 24.0, 32.0],
    'Smoking Status':[1,0,1],
    'Alcohol Intake':[0,1,1],
    'Physical Activity':[1,0,0],
    'Stroke History':[0,0,1],
    'Family History of Stroke':[1,0,1],
    'Dietary Habits':[1,0,1],
    'Stress Levels':[0,1,2],
    'Symptoms':[1,0,1],
    'HDL_Cholesterol':[55,60,45],
    'LDL_Cholesterol':[120,90,160],
    'Systolic_BP':[124, 118, 150],
    'Diastolic_BP':[80, 75, 95]
})

# Ensure columns match X_train
new_data = new_data[X_train.columns]

# Scale the new data (for SVM / Linear Regression)
new_data_scaled = scaler.transform(new_data)

# Predict probability using the best model
if best_model_name == "Random Forest":
    new_prob = rf_model.predict_proba(new_data_scaled)[:,1]
elif best_model_name == "SVM":
    new_prob = svm_model.predict_proba(new_data_scaled)[:,1]
else:
    new_prob = lr_model.predict(new_data_scaled)
    # Clip to [0,1] range
    new_prob = [max(0, min(1, x)) for x in new_prob]

# Convert to class prediction using threshold 0.3
new_pred_class = [1 if p >= 0.5 else 0 for p in new_prob]

# Display results
df_new_pred = new_data.copy()
df_new_pred['Predicted_Probability'] = new_prob
df_new_pred['Predicted_Stroke'] = new_pred_class

print("\nPredicted Stroke Probability and Class for New Data:")
display(df_new_pred)
```
