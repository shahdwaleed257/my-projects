# Numbers 

## Types of numbers

We'll mainly focus on integers and floating point numbers.

Integers are just whole numbers, positive or negative. For example: 2 and -2 are examples of integers.

Floating point numbers in Python are notable because they have a decimal point in them, or use an exponential (e) to define the number. For example 2.0 and -2.1 are examples of floating point numbers. 
4E2 (4 times 10 to the power of 2) is also an example of a floating point number in Python.

# Comments 
Comments are ways to write extra text for clarification of your code: 
- Single Line Comments: Uses # to define a comment in only one line 
- Multi-Line Comment: Uses """ multi-line comment """ for comments spanning more than one line.  

```python
a = 2
b = 3
# Single Line Comment
# Operation for c
c = a+(b*3)

"""
Multi-Line Comment

c: refers to the result I need to calculate
a: value equals 2 
b: value eqals 3
"""
print(c)
```

Yes! Python allows you to write over assigned variable names. We can also use the variables themselves when doing the reassignment. Here is an example of what I mean:

```python
a = 2.5
b = 2
print(a*b)

# Change the value of b from 2 to 2.2
b = 2.2
print(a*b,round(a*b,0))
```

The names you use when creating these labels need to follow a few rules:

    1. Names can not start with a number.
    2. There can be no spaces in the name, use _ instead.
    3. Can't use any of these symbols :'",<>/?|\()!@#$%^&*~-+
    4. It's considered best practice (PEP8) that names are lowercase.
    5. Avoid using the characters 'l' (lowercase letter el), 'O' (uppercase letter oh), 
       or 'I' (uppercase letter eye) as single character variable names.
    6. Avoid using words that have special meaning in Python like "list" and "str"
