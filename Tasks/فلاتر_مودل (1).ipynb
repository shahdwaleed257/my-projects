```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
import joblib
import tensorflow as tf

# 1. إعداد البيانات
data = {'حمى': [1, 0, 1, 0, 1],
        'صداع': [1, 1, 0, 0, 1],
        'إرهاق': [1, 1, 1, 0, 1],
        'آلام بطن': [0, 0, 1, 1, 0],
        'مرض': ['إنفلونزا', 'صداع نصفي', 'تسمم غذائي', 'التهاب معدة', 'إنفلونزا']}

df = pd.DataFrame(data)

# 2. تحديد المدخلات والمخرجات
X = df[['حمى', 'صداع', 'إرهاق', 'آلام بطن']]  # الأعراض
y = df['مرض']  # التشخيص

# 3. تدريب نموذج Decision Tree
model = DecisionTreeClassifier()
model.fit(X, y)

# 4. حفظ النموذج باستخدام Joblib
joblib.dump(model, 'decision_tree_model.pkl')
print("✅ تم حفظ نموذج Decision Tree!")

# 5. تحويل النموذج إلى TFLite
def convert_to_tflite():
    import numpy as np

    # تحويل النموذج إلى نموذج TensorFlow
    inputs = tf.keras.Input(shape=(4,), dtype=tf.float32)
    outputs = tf.keras.layers.Dense(3, activation="softmax")(inputs)
    tf_model = tf.keras.Model(inputs=inputs, outputs=outputs)

    # حفظ النموذج
    converter = tf.lite.TFLiteConverter.from_keras_model(tf_model)
    tflite_model = converter.convert()

    with open('decision_tree_model.tflite', 'wb') as f:
        f.write(tflite_model)
    print("✅ تم تحويل النموذج إلى TFLite!")

convert_to_tflite()
# حفظ النموذج المحول إلى TFLite
with open('/content/medical_diagnosis_model.tflite', 'wb') as f:
    f.write(tflite_model)

# لتحميل الملف إلى جهازك المحلي
from google.colab import files
files.download('/content/medical_diagnosis_model.tflite')
```
